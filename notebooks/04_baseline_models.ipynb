```python
import torch
from src.baseline_models import B0_Geometric, B1_GPSOnly, B2_RGBOnly
from src.fusion_models import B3_Fusion, B4_MultimodalTransformer

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Device: {device}')

# Dummy batch
rgb = torch.randn(2, 5, 3, 96, 96).to(device)
gps = torch.randn(2, 5, 9).to(device)

for name, ModelClass in [('B0', B0_Geometric), ('B1', B1_GPSOnly),
                          ('B2', B2_RGBOnly), ('B3', B3_Fusion),
                          ('B4', B4_MultimodalTransformer)]:
    m = ModelClass().to(device)
    out = m(rgb, gps)
    n_params = sum(p.numel() for p in m.parameters()) / 1e6
    print(f'  {name}: logits={out["logits"].shape}  params={n_params:.2f}M')
```
